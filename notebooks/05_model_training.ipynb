```python
import pandas as pd
import numpy as np

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
df = pd.read_csv("../data/processed/model_data.csv")
```

```python
# convert date 
df["Date"] = pd.to_datetime(df["Date"])
```

```python
# sort by date 
df = df.sort_values("Date").reset_index(drop=True)
```

```python
# train \test split 
train_size = int(len(df) * 0.8)

train = df.iloc[:train_size]
test = df.iloc[train_size:]
```

```python
# feature  and target 
X_train = train.drop(columns=["Sales", "Date"], errors="ignore")
y_train = train["Sales"]

X_test = test.drop(columns=["Sales", "Date"], errors="ignore")
y_test = test["Sales"]
```

```python
# categorical features 
categorical_cols = [
    "Product_ID",
    "Product",
    "Category",
    "Store",
    "City"
]
```

```python
# preprocessing 
preprocessor = ColumnTransformer(
    transformers=[
        (
            "cat",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_cols
        )
    ],
    remainder="passthrough"
)
```

```python
# model pipleline
model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("regressor", LinearRegression())
    ]
)
```

```python
# model training 
model.fit(X_train, y_train)

# prediction 
y_pred = model.predict(X_test)
```

```python
# evaluation 
mae = mean_absolute_error(y_test, y_pred)

rmse = np.sqrt(
    mean_squared_error(y_test, y_pred)
)

r2 = r2_score(y_test, y_pred)


print("Model: Linear Regression")
print("------------------------")
print("MAE :", mae)
print("RMSE:", rmse)
print("R²  :", r2)
```

```python
import pandas as pd
import numpy as np

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
# Load data
df = pd.read_csv("../data/processed/model_data.csv")

# Date + sorting
df["Date"] = pd.to_datetime(df["Date"])
df = df.sort_values("Date").dropna().reset_index(drop=True)
```

```python
# Train/Test split
split = int(len(df) * 0.8)
train = df.iloc[:split]
test = df.iloc[split:]

# Features and target
X_train = train.drop(columns=["Sales", "Date"])
y_train = train["Sales"]

X_test = test.drop(columns=["Sales", "Date"])
y_test = test["Sales"]
```

```python
# Categorical columns
cat_cols = ["Product_ID", "Product", "Category", "Store", "City"]

# Preprocessing
preprocessor = ColumnTransformer(
    [("cat", OneHotEncoder(handle_unknown="ignore"), cat_cols)],
    remainder="passthrough"
)
```

```python
model = Pipeline([
    ("preprocessor", preprocessor),
    ("rf", RandomForestRegressor(
        n_estimators=10,
        max_depth=8,
        random_state=42,
        n_jobs=-1
    ))
])

model.fit(X_train, y_train)
y_pred = model.predict(X_test)
```

```python
import joblib

joblib.dump(model, "../models/random_forest_model.pkl")

print("Model saved successfully!")
```

```python
# Evaluate
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("Random Forest")
print("MAE :", mae)
print("RMSE:", rmse)
print("R²  :", r2)
```
